# TyphoFormer-for-Hurricane-Track-Forecasting: offline results explorer
This notebook clones the public source and displays a browser-only viewer of committed results. No model weights, GPU, paid APIs or new experimental claims.


```python
import subprocess
from pathlib import Path
from IPython.display import HTML, display
repo = Path("TyphoFormer-for-Hurricane-Track-Forecasting")
if not repo.exists():
    subprocess.run(["git", "clone", "--depth", "1", "https://github.com/WYH302/TyphoFormer-for-Hurricane-Track-Forecasting.git", str(repo)], check=True)
subprocess.run(["python", "demo.py", "--no-open"], cwd=repo, check=True)
display(HTML((repo / "demo_output/index.html").read_text(encoding="utf-8")))
```
